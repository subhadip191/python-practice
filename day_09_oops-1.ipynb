### Constructors

constructor is a special method that automatically runs when an object is created.

````python
class Temp:

  def __init__(self):
    print('hello')

obj = Temp()
````

````python
class Atm:
    def __init__(self):
        print(id(self))
        self.balance = 0

    def check_balance(self):
        print("Balance:", self.balance)

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")
````

````python
atm1 = Atm()
id(atm1)
````

````python
atm2 = Atm()
id(atm2)
````

````python
L = [1,2,3]
len(L) # function -> because it is outside the list class
L.append(4) # method -> because it is inside the list class
````

````python
# methods
atm1.check_balance()
atm2.deposit(500)
atm1.withdraw(300)
atm2.check_balance()
````

### Magic Methods

Magic methods are special methods in Python that start and end with double underscores (`__`).
They allow us to define how an object behaves when we use built-in Python operations or functions with it.
These methods are also called **dunder methods** (double-underscore methods).



````python
class ShoppingCart:

    def __init__(self, owner):
        self.owner = owner
        self.items = []

    def add_item(self, item, price):
        self.items.append((item, price))

    # 1. __str__ → controls print(object)
    def __str__(self):
        return f"{self.owner}'s Cart - {len(self.items)} items"

    # 2. __len__ → controls len(object)
    def __len__(self):
        return len(self.items)

    # 3. __getitem__ → allows object[index]
    def __getitem__(self, index):
        return self.items[index]

    # 4. __contains__ → allows item in object
    def __contains__(self, item):
        return any(name == item for name, price in self.items)

    # 5. __add__ → allows cart1 + cart2
    def __add__(self, other):
        new_cart = ShoppingCart(self.owner)
        new_cart.items = self.items + other.items
        return new_cart
````

````python
cart1 = ShoppingCart("Subhadip")

cart1.add_item("Laptop", 1000)
cart1.add_item("Mouse", 50)
cart1.add_item("Keyboard", 80)
````

````python
# __str__
print(cart1)

# __len__
print(len(cart1))

# __getitem__
print(cart1[0])

# __contains__
print("Mouse" in cart1)
````

````python
cart2 = ShoppingCart("Friend")
cart2.add_item("Monitor", 300)

# __add__
cart3 = cart1 + cart2

print(cart3)
print(len(cart3))
````
