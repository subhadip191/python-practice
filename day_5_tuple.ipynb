# Tuples

A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.

### Characterstics

- Ordered
- Unchangeble
- Allows duplicate

### Plan of attack

- Creating a Tuple
- Accessing items
- Editing items
- Adding items
- Deleting items
- Operations on Tuples
- Tuple Functions

### Creating Tuples

````python
# empty
t1 = ()
print(t1)
# create a tuple with a single item
t2 = ('hello',)
print(t2)
print(type(t2))
# homo
t3 = (1,2,3,4)
print(t3)
# hetro
t4 = (1,2.5,True,[1,2,3])
print(t4)
# tuple
t5 = (1,2,3,(4,5))
print(t5)
# using type conversion
t6 = tuple('hello')
print(t6)
````

### Accessing Items

### Operations on Tuples

````python
# + and *
t1 = (1,2,3,4)
t2 = (5,6,7,8)

print(t1 + t2)

print(t1*3)

# membership
print(1 in t1)

# iteration
for i in t1:
  print(i)
````

### Tuple Functions

````python
# len/sum/min/max/sorted
t = (1,2,3,4)

print(len(t))

print(sum(t))

print(min(t))

print(max(t))

sorted(t,reverse=True)
````

### Difference between Lists and Tuples

- Syntax

- Speed
- Memory
- Built in functionality
- Error prone
- Usability


````python
import time

L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*5
print('List time',time.time()-start)

start = time.time()
for i in T:
  i*5
print('Tuple time',time.time()-start)
````

````python
import sys

L = list(range(1000))
T = tuple(range(1000))

print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
````

### `Q1:` Join Tuples if similar initial element
While working with Python tuples, we can have a problem in which we need to perform concatenation of records from the similarity of initial element. This problem can have applications in data domains such as Data Science.

For eg.
```
Input  : test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
Output : [(5, 6, 7, 8), (6, 10), (7, 13)]
```

````python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

new = []

seen = float('-inf')
temp = ()

for i in range(len(test_list)):
    if test_list[i][0] != seen:
        if temp:
            new.append(temp)
        seen = test_list[i][0]
        temp = test_list[i]
    else:
        temp += (test_list[i][1],)

new.append(temp)
print(new)
````

### `Q2`: Check is tuples are same or not?
Two tuples would be same if both tuples have same element at same index
```
t1 = (1,2,3,0)
t2 = (0,1,2,3)

t1 and t2 are not same
```

````python
t1 = (1,2,3,0)
t2 = (0,1,2,3)
t1 == t2
````

````python
same = True

if len(t1) != len(t2):
    same = False
else:
    for i in range(len(t1)):
        if t1[i] != t2[i]:
            same = False
            break

print(same)
````

### `Q3`: Count no of tuples, list and set from a list
```
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]

```
`Output:`

```
List-2
Set-2
Tuples-1
```

````python
list1 = [{'hi', 'bye'},{'Geeks', 'forGeeks'},('a', 'b'),['hi', 'bye'],['a', 'b']]
dis = {'List':0, 'Set':0, 'Tuple':0}
for i in list1:
    if isinstance(i, list):
        dis['List'] += 1
    elif isinstance(i, set):
        dis['Set'] += 1
    else:
        dis['Tuple'] += 1

print(f'List - {dis["List"]}')
print(f'Set - {dis["Set"]}')
print(f'Tuple - {dis["Tuple"]}')
````
