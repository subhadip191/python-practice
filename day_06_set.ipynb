# Sets

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.

Characterstics:
- Unordered
- Mutable
- No Duplicates
- Can't contain mutable data types

### Creating Sets

````python
# empty
s = set()
print(s)

print(type(s))

# 1D and 2D
s1 = {1,2,3}
print(s1)

s2 = {1,2,3,{4,5}}
````

````python
# homo and hetro
s3 = {1,'hello',4.5,(1,2,3)}
print(s3)

# using type conversion
s4 = set([1,2,3])
print(s4)

# duplicates not allowed
s5 = {1,1,2,2,3,3}
print(s5)

# set can't have mutable items
s6 = {1,2,[3,4]}
````

````python
s1 = {1,2,3}
s2 = {3,2,1}

print(s1 == s2)
````

### Set Operation

````python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# Union(|)
print(s1 | s2)

# Intersection(&)
print(s1 & s2)

# Difference(-)
print(s1 - s2)
print(s2 - s1)

# Symmetric Difference(^)
print(s1 ^ s2)

# Membership
1 not in s1

# Iteration
for i in s1:
  print(i)
````

### Set Functions

````python
# len/sum/min/max/sorted
s = {3,1,4,5,2,7}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
sorted(s,reverse=True)
````

````python
# union/update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# s1 | s2
s1.union(s1)
print(s1)
print(s2)

s1.update(s2)
print(s1)
print(s2)
````

````python
# intersection/intersection_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)
print(s1)
print(s2)

s1.intersection_update(s2)
print(s1)
print(s2)
````

````python
# difference/difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)
print(s1)
print(s2)

s1.difference_update(s2)
print(s1)
print(s2)
````

````python
# symmetric_difference/symmetric_difference_update
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2)
print(s1)
print(s2)

s1.symmetric_difference_update(s2)
print(s1)
print(s2)
````

````python
# isdisjoint/issubset/issuperset
s1 = {1,2,3,4}
s2 = {7,8,5,6}

s1.isdisjoint(s2)
````

````python
s1 = {1,2,3,4,5}
s2 = {3,4,5}

s1.issuperset(s2)
````

````python
# copy
s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
````

# Set Problem

### `Q1:` Write a program to find set of common elements in three lists using sets.
```
Input : ar1 = [1, 5, 10, 20, 40, 80]
        ar2 = [6, 7, 20, 80, 100]
        ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

Output : [80, 20]
```

````python
ar1 = [1, 5, 10, 20, 40, 80]
ar2 = [6, 7, 20, 80, 100]
ar3 = [3, 4, 15, 20, 30, 70, 80, 120]

x = set(ar1)
y = set(ar2)
z = set(ar3)

x.intersection_update(y)
x.intersection_update(z)
print(x)
````

### `Q2:` Write a program to count unique number of vowels using sets in a given string. Lowercase and upercase vowels will be taken as different.

`Input:`
```
Str1 = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
```
`Output:`
```
No of unique vowels-6
```

````python
s = "hands-on data science mentorship progrAm with live classes at affordable fee only on CampusX"
unique = set()
vowels = {'a', 'e', 'i', 'o', 'u'}
for i in s:
    if i.lower() in vowels:
        unique.add(i)

print('No of unique vowels- ',len(unique))

````

### `Q3:` Write a program to Check if a given string is binary string of or not.

A string is said to be binary if it's consists of only two unique characters.

Take string input from user.

```
Input: str = "01010101010"
Output: Yes

Input: str = "1222211"
Output: Yes

Input: str = "Campusx"
Output: No
```

````python
N = input()
s = set()

for i in N:
    s.add(i)

if len(s) == 2:
    print('Yes')
else:
    print('False')
````

### `Q4`: find union of n arrays.

**Example 1:**

Input:
```bash
[[1, 2, 2, 4, 3, 6],
 [5, 1, 3, 4],
 [9, 5, 7, 1],
 [2, 4, 1, 3]]
```

Output:

```bash
[1, 2, 3, 4, 5, 6, 7, 9]
```

````python
l = [[1, 2, 2, 4, 3, 6], [5, 1, 3, 4], [9, 5, 7, 1], [2, 4, 1, 3]]
union = set()

for i in range(0, len(l)):
    union.update(set(l[i]))

print(list(union))
````

### `Q5`: Intersection of two lists. Intersection of two list means we need to take all those elements which are common to both of the initial lists and store them into another list. Only use using **list-comprehension**.

**Example 1:**

Input:
```bash
lst1 = {15, 9, 10, 56, 23, 78, 5, 4, 9}
lst2 = {9, 4, 5, 36, 47, 26, 10, 45, 87}
```

Output:
```bash
[9, 10, 4, 5]
```

**Example 2:**

Input:
```bash
lst1 = {4, 9, 1, 17, 11, 26, 28, 54, 69}
lst2 = {9, 9, 74, 21, 45, 11, 63, 28, 26}
```

Output:
```bash
[9, 11, 26, 28]
```

````python
lst1 = {15, 9, 10, 56, 23, 78, 5, 4, 9}
lst2 = {9, 4, 5, 36, 47, 26, 10, 45, 87}

intersection = {i for i in lst1 if i in lst2 }
print(intersection)
````
