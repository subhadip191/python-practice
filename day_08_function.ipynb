### `*args and **kwargs`
`*args and **kwargs` are special Python keywords that are used to pass the variable length of arguments to a function

````python
# *args
# allows us to pass a variable number of non-keyword arguments to a function.

def multiply(*args):
  product = 1

  for i in args:
    product = product * i

  print(args)
  return product
````

````python
multiply(1,2,3,4,5,6,7,8,9)
````

````python
# **kwargs
# **kwargs allows us to pass any number of keyword arguments.
# Keyword arguments mean that they contain a key-value pair, like a Python dictionary.

def display(**inp):

  for (key,value) in inp.items():
    print(key,'->',value)
  print(inp)
````

````python
display(india = 'delhi', italy = 'rome', france = 'paris')
````

##### Points to remember while using `*args and **kwargs`

- order of the arguments matter(normal -> `*args` -> `**kwargs`)
- The words “args” and “kwargs” are only a convention, you can use any name of your choice


````python
def power_of_element(power, *args):
  lst =[]

  for i in range(len(args)):
    lst.append(args[i]**power)

  return lst

power_of_element(3,1,2,3,4,5,6,7)
````

### Lambda Function

A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

````python
# x -> x^2
lambda x:x**2
````

````python
# x,y -> x+y
a = lambda x,y:x+y
a(5,2)
````

#### Diff between lambda vs Normal Function

- No name
- lambda has no return value(infact,returns a function)
- lambda is written in 1 line
- not reusable

Then why use lambda functions?<br>
**They are used with HOF**


````python
# check if a string has 'a'
a = lambda s:'a' in s
a('hello')
````

````python
# odd or even
a = lambda x:'even' if x%2 == 0 else 'odd'
a(6)
````

### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
def sort_str(s):
    lis = s.split('-')
    lis.sort()
    out = ''
    for i in lis:
        out += i + '-'

    return out

s = 'green-red-yellow-black-white'

sort_str(s)
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
def count(s):
    tracker = {'upper':0, 'lower':0}
    for i in range(len(s)):
        if s[i].isupper():
            tracker['upper'] += 1
        if s[i].islower():
            tracker['lower'] += 1

    return tracker

s = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
x = count(s)

print(f"No. of upper case char: {x['upper']}")
print(f"No. of lower case char: {x['lower']}")
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
def perfect_num(num):
    add = 0

    for i in range(1, num // 2 + 1):
        if num % i == 0:
            add += i

    return add == num

perfect_num(28)
````

### **`Problem-6:`** Write a Python function to concatenate any no of dictionaries to create a new one.

```
Sample Dictionary :
dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
Expected Result : {1: 10, 2: 20, 3: 30, 4: 40, 5: 50, 6: 60}
```

````python
def con(*args):
    dic = {}
    for i in args:
        for key, val in i.keys(),i.values():
            dic[key] = val
    return dic

dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}
con(dic1, dic2, dic3)
````

### `Problem-7` Write a python function that accepts a string as input and returns the word with most occurence.

```
Input:
hello how are you i am fine thank you
```

```
Output
you -> 2
```

````python
def word_occ(s):
    dic = {}
    for i in s.split(' '):
        if i in dic:
            dic[i] += 1
        else:
            dic[i] = 1

    return f"{max(dic, key=dic.get)} -> {dic[max(dic, key=dic.get)]}"

s = 'hello how are you i am fine thank you'

word_occ(s)
````

### `Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
def hist(lis, bucket = 10):
    lis.sort()
    low = 1
    high = bucket
    dic = {}
    for i in lis:
        while i > high:
            low = high+1
            high += bucket
        k = f'{low}-{high}'
        if k not in dic:
            dic[k] = 1
        else:
            dic[k] += 1
    return dic

lis = [13,42,15,37,22,39,41,50]

hist(lis)
````

### `Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
def word(s):
    dic = {}
    s = s.lower()

    for i in ".,!?;:":
        s = s.replace(i, "")

    for i in s.split(' '):
        if i in dic:
            dic[i] += 1
        else:
            dic[i] = 1
    return dic

s = "John likes to watch movies. Mary likes movies too."

word(s)
````

### `Problem 11:` Write a Python program to add three given lists using Python map and lambda.

````python
def add_list(*args):
    length = max(map(len, args))
    result = []
    for i in range(length):
        result.append(sum(map(lambda x: x[i] if i < len(x) else 0, args)))
    return result

l1 = [1,2,3,4,5,6,7]
l2 = [9,8,7,6,5,4,3]
l3 = [11,22,33,44,55]

add_list(l1,l2,l3)

````

### `Problem-12:`Write a Python program to create a list containing the power of said number in bases raised to the corresponding number in the index using Python map.
`Input:`
```
list1 = [1,2,3,4,5,6]
```
`Output:`
```
[1,2,9,64,625,-]
```


````python
def power_list(l):
    result = l[:1]
    result.extend(
        map(lambda i: l[i] ** l[i - 1], range(1, len(l)-1))
    )
    result.append('-')
    return result

l = [1,2,3,4,5,6]
power_list(l)
````

### `Problem-13` Using filter() and list() functions and .lower() method filter all the vowels in a given string.



````python
def filter_vowels(s):
    vowels = "aeiou"
    return "".join(filter(lambda ch: ch.lower() not in vowels, s))

s = "Hello  everyone my name is Subhadip Maity, and I am a DS student"
filter_vowels(s)
````

### `Problem-14`: Use reduce to convert a 2D list to 1D

````python
import functools
def convert(l):
    lis = []
    lis.extend(functools.reduce(lambda x,y: x+y, l))
    return lis

l = [[1,2,3,4], [5,6,7,8]]
convert(l)
````
