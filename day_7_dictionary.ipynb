# Dictionary

Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }

Characterstics:

- Mutable
- Indexing has no meaning
- keys can't be duplicated
- keys can't be mutable items

### Create Dictionary

````python
# empty dictionary
d = {}
print(d)

# 1D dictionary
d1 = { 'name' : 'Subhadip' ,'gender' : 'male' }
print(d1)

# with mixed keys
d2 = {(1,2,3):1,'hello':'world'}
print(d2)

# 2D dictionary -> JSON
s = {
    'name':'subhadip',
     'college':'unina',
     'sem':3,
     'subjects':{
         'dsa':60,
         'maths':65,
         'english':55
     }
}

print(s)
````

````python
# using sequence and dict function
d4 = dict([('name','subhadip'),('age',22),(3,3)])
print(d4)

# duplicate keys
d5 = {'name':'subhadip','name':'as'}
print(d5)

# mutable items as keys
d6 = {'name':'nitish',[1,2,3]:2}
````

### Accessing items

````python
my_dict = {'name': 'Jack', 'age': 26}

# []
print(my_dict['age'])

# get
print(my_dict.get('age'))

# access from 2d
print(s['subjects']['maths'])
````

### Adding key-value pair

````python
d4['gender'] = 'male'
print(d4)

s['subjects']['ds'] = 75
s
````

### Remove key-value pair

````python
d = {'name': 'subhadip', 'age': 22, 3: 3, 'gender': 'male', 'weight': 80}
# pop
d.pop(3)
print(d)

# popitem
d.popitem()
print(d)

# del
del d['name']
print(d)

# clear
d.clear()
print(d)

del s['subjects']['maths']
print(s)
````

### Editing key-value pair

````python
s['subjects']['dsa'] = 80
s
````

### Dictionary Comprehension

{ key : value for vars in iterable condition }

````python
# print 1st 10 numbers and their squares
{i:i**2 for i in range(1,11)}
````

````python
distances = {'delhi':1000,'mumbai':2000,'bangalore':3000}
print(distances.items())
````

````python
# using existing dict
distances = {'delhi':1000,'mumbai':2000,'bangalore':3000}
{key:value*0.62 for (key,value) in distances.items()}
````

````python
# using zip
days = ["Sunday", "Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
temp_C = [30.5,32.6,31.8,33.4,29.8,30.2,29.9]

{i:j for (i,j) in zip(days,temp_C)}
````

````python
# using if condition
products = {'phone':10,'laptop':0,'charger':32,'tablet':0}

{key:value for (key,value) in products.items() if value>0}
````

````python
# Nested Comprehension
# print tables of number from 2 to 4
{i:{j:i*j for j in range(1,11)} for i in range(2,5)}
````

### `Q1`: Key with maximum unique values

Given a dictionary with values list, extract key whose value has most unique values.

**Example 1:**

Input:

```bash
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
CampusX
```

**Example 2:**

Input:
```bash
test_dict = {"CampusX" : [5, 7, 7, 7, 7], "is" : [6, 7, 7, 7], "Best" : [9, 9, 6, 5, 5]}
```

Output:
```bash
Best
```

````python
test_dict = {"CampusX" : [5, 7, 9, 4, 0], "is" : [6, 7, 4, 3, 3], "Best" : [9, 9, 6, 5, 5]}
max_value = 0
max_key = None
for i in test_dict.keys():
    unique = len(set(test_dict[i]))
    if unique > max_value:
        max_value = unique
        max_key = i

print(max_key)
````

### `Q2`: Replace words from Dictionary. Given String, replace it’s words from lookup dictionary.

**Example 1:**

Input:

```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}
```

Output:

```bash
CampusX is the best channel for Data-Science students.
```

**Example 2:**

Input:
```bash
test_str = 'CampusX best for DS students.'
repl_dict = {"good" : "is the best channel", "ds" : "Data-Science"}
```

Output:
```bash
CampusX best for DS students.
```

````python
test_str = 'CampusX best for DS students.'
repl_dict = {"best" : "is the best channel", "DS" : "Data-Science"}

string = ''

for i in test_str.split(' '):
    if i in repl_dict:
        string += repl_dict[i]
    else:
        string += i

    string += ' '

print(string)
````

### `Q3`: Convert List to List of dictionaries. Given list values and keys list, convert these values to key value pairs in form of list of dictionaries.

**Example 1:**

Input:
```bash
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'DataScience', 'id': 3}, {'name': 'is', 'id': 8}]
```

**Example 2:**

Input:
```bash
test_list = ["CampusX", 10]
key_list = ["name", "id"]
```

Output:

```bash
[{'name': 'CampusX', 'id': 10}]
```

````python
test_list = ["DataScience", 3, "is", 8]
key_list = ["name", "id"]
d = {}
lis = []
for i in range(len(test_list)):
    d[key_list[i%len(key_list)]] = test_list[i]
    if len(d) == len(key_list):
        lis.append(d)
        d = {}

print(lis)
````

### `Q4`: Convert a list of Tuples into Dictionary.

**Example 1:**

Input:
```bash
[("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
```

Output:
```bash
{'akash': [10], 'gaurav': [12], 'anand': [14], 'suraj': [20], 'akhil': [25], 'ashish': [30]}
```

**Example 2:**

Input:
```bash
[('A', 1), ('B', 2), ('C', 3)]
```

Output:
```bash
{'A': [1], 'B': [2], 'C': [3]}
```

````python
lst =  [("akash", 10), ("gaurav", 12), ("anand", 14), ("suraj", 20), ("akhil", 25), ("ashish", 30)]
dic = {}
for i in range(len(lst)):
    dic[lst[i][0]] = [lst[i][1]]

print(dic)
````

### `Q5`: Sort Dictionary key and values List.

**Example 1:**

Input:

```bash
{'c': [3], 'b': [12, 10], 'a': [19, 4]}
```

Output:

```bash
{'a': [4, 19], 'b': [10, 12], 'c': [3]}
```

**Example 2:**

Input:

```bash
{'c': [10, 34, 3]}
```

Output:

```bash
{'c': [3, 10, 34]}
```

````python
d = {'c': [3], 'b': [12, 10], 'a': [19, 4]}
sorted_key = dict(sorted(d.items()))

for i in sorted_key.keys():
    sorted_key[i] = sorted(sorted_key[i])

print(sorted_key)
````
